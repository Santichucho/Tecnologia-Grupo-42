### 1. La Identidad de los Datos (`type`)
Python tiene una función "inspector" llamada `type()` que nos dice qué es exactamente un dato.

Ejecuta el siguiente bloque y observa la diferencia entre el 10, el 10.5 y el "10".

```python
entero = 10
decimal = 10.5
texto = "10"
logico = True

print("El valor", entero, "es de tipo:", type(entero))
print("El valor", decimal, "es de tipo:", type(decimal))
print("El valor", texto, "es de tipo:", type(texto))
print("El valor", logico, "es de tipo:", type(logico))
```

### 2. Matemáticas de Ingeniería: Divisiones y Resto
Cualquiera puede sumar. Un ingeniero debe saber dividir correctamente.

* `/` División normal (siempre da decimales).
* `//` División entera (corta los decimales).
* `%` **Módulo** (nos dice cuánto sobra).

**Ejemplo de la Pizzería:**
Tenemos 10 porciones de pizza y 3 estudiantes. ¿Cuántas come cada uno y cuántas sobran?

```python
porciones = 10
estudiantes = 3

# División Entera (//)
comen_cada_uno = porciones // estudiantes

# Módulo (%)
sobran = porciones % estudiantes

print("Cada estudiante come:", comen_cada_uno, "porciones.")
print("Sobran en la caja:", sobran, "porciones.")
```

### 3. El "Crash Test": Sumar Texto vs. Sumar Números
Este es el error más común en los exámenes de ingreso.
La función `input()` **SIEMPRE devuelve Texto (string)**, aunque el usuario escriba un número.

Ejecuta el bloque, ingresa `100` y luego `50`. Mira el desastre.

```python
print("--- Calculadora Rota ---")
dato1 = input("Ingresa un número: ")
dato2 = input("Ingresa otro número: ")

resultado = dato1 + dato2

print("La suma es:", resultado)
```

**¿Te dio 10050?** 🚨
Eso es porque el operador `+` en texto hace **Concatenación** (pegar), no suma matemática.

### 4. La Solución: Casting (Conversión)
Para arreglar esto, debemos transformar el dato usando `int()` o `float()`.

Arreglemos la calculadora:

```python
print("--- Calculadora DE INGENIERO (Corregida) ---")

# Paso 1: Pedimos texto
texto1 = input("Ingresa el primer precio: ")
texto2 = input("Ingresa el segundo precio: ")

# Paso 2: Convertimos (Casting) a Decimales (float) por si tiene centavos
num1 = float(texto1)
num2 = float(texto2)

# Paso 3: Operamos matemáticamente
total = num1 + num2

print("El total real a pagar es:", total)
```
