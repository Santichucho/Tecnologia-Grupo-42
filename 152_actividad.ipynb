# Modularidad y Funciones

En este bloque aprenderás por qué las funciones son esenciales para la programación y cómo la **modularidad** ayuda a resolver problemas de forma organizada.


### Ventajas de usar funciones
Las funciones permiten descomponer problemas complejos en piezas más pequeñas. Entre sus ventajas principales se encuentran:
- **Simplificación de la complejidad**: dividir un problema grande en sub‑tareas facilita entenderlo y abordarlo.
- **Mantenimiento y flexibilidad**: modificar la lógica de una función no afecta al resto del programa.
- **Reutilización**: el mismo bloque puede emplearse en distintos contextos sin reescribirlo.
- **Trabajo en equipo**: cada desarrollador puede encargarse de diferentes funciones sin interferencias.

Además, organizar el código en funciones favorece la **modularidad**, es decir, la construcción de un sistema a partir de módulos independientes. La modularidad hace que el código sea más legible y escalable.


### Definición básica de una función en Python
En Python, una función se define con la palabra clave `def`, seguida del nombre, una lista de parámetros entre paréntesis y un bloque de código indentado. Todas las funciones devuelven un valor de manera explícita (`return`) o implícita (`None`).
```python
def saludar():
    print("¡Hola!")

saludar()  # Llama a la función
```

La función `saludar` no recibe parámetros y solo imprime un mensaje. Aunque no tenga `return`, Python devolverá automáticamente `None`.


### Parámetros y argumentos
Los **parámetros** son las variables que aparecen en la definición de una función; los **argumentos** son los valores reales que se pasan al invocar la función.

#### Alcance de las variables
Los parámetros y variables definidos dentro de una función son **locales**. No afectan a las variables externas a menos que se modifique un objeto mutable (por ejemplo, una lista o diccionario).

Veamos ejemplos que muestran la diferencia entre pasar un número (inmutable) y una lista (mutable):


````python
# Parámetros inmutables
def incrementa_numero(n):
    n = n + 1
    print(f"Dentro de la función: n = {n}")

x = 10
incrementa_numero(x)
print(f"Fuera de la función: x = {x}")

# Parámetros mutables
def agrega_elemento(lista):
    lista.append("nuevo")
    print(f"Dentro de la función: lista = {lista}")

mi_lista = ['a', 'b']
agrega_elemento(mi_lista)
print(f"Fuera de la función: lista = {mi_lista}")
````

En el caso de un entero (`x`), el valor fuera de la función no cambia porque los números son inmutables.
Sin embargo, al pasar una lista, la función modifica la lista original porque las listas son objetos mutables.


### Funciones que llaman otras funciones
Una función puede invocar a otra para reutilizar su lógica. Esto potencia la modularidad y evita duplicar código.

Por ejemplo, definamos una función `cuadrado()` que calcula el cuadrado de un número y otra función `suma_cuadrados()` que utilice `cuadrado()` para sumar los cuadrados de dos números.


````python
def cuadrado(n):
    return n * n

def suma_cuadrados(a, b):
    return cuadrado(a) + cuadrado(b)

resultado = suma_cuadrados(3, 4)
print(f"La suma de los cuadrados de 3 y 4 es {resultado}")
````

En este caso, `suma_cuadrados` reutiliza la lógica de `cuadrado` para mantener el código limpio y fácil de mantener.


### Alcance (scope) y valores de retorno
Las variables definidas dentro de una función son **locales**, es decir, solo existen dentro del bloque de la función.
Las funciones devuelven valores mediante la palabra clave `return`. Una función sin `return` explícito devuelve `None`.
También es posible devolver varios valores a la vez utilizando una tupla.


````python
def operaciones(a, b):
    suma = a + b
    resta = a - b
    multiplicacion = a * b
    return suma, resta, multiplicacion

# Llamar a la función y desempaquetar los resultados
s, r, m = operaciones(8, 3)
print("Suma =", s)
print("Resta =", r)
print("Multiplicación =", m)
````

### Descomponer un problema complejo en funciones
Supongamos que queremos calcular el índice de masa corporal (IMC) de varias personas y clasificarlas como 'bajo peso', 'normal', 'sobrepeso' u 'obesidad'.
En lugar de escribir un único bloque de código largo, podemos crear funciones para cada tarea.

- `calcular_imc(peso, altura)`: devuelve el IMC.
- `clasificar_imc(imc)`: devuelve una categoría según el IMC.
- `procesar_pacientes(pacientes)`: recorre la lista de pacientes (tuplas de nombre, peso y altura), calcula su IMC y clasificación y devuelve una lista de resultados.


````python
def calcular_imc(peso, altura):
    # altura en metros
    return peso / (altura ** 2)

def clasificar_imc(imc):
    if imc < 18.5:
        return "Bajo peso"
    elif 18.5 <= imc < 25:
        return "Normal"
    elif 25 <= imc < 30:
        return "Sobrepeso"
    else:
        return "Obesidad"

def procesar_pacientes(pacientes):
    resultados = []
    for nombre, peso, altura in pacientes:
        imc = calcular_imc(peso, altura)
        categoria = clasificar_imc(imc)
        resultados.append((nombre, round(imc, 1), categoria))
    return resultados

# Lista de personas (nombre, peso en kg, altura en metros)
personas = [
    ("Ana", 50, 1.60),
    ("Luis", 80, 1.75),
    ("María", 70, 1.68)
]

for persona in procesar_pacientes(personas):
    print(persona)
````
