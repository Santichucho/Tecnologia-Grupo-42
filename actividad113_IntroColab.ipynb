### 2. El Concepto de "Memoria" (Variables)
Aquí empieza la programación real.
Si solo escribimos `2 + 2`, la computadora lo calcula y lo **olvida** al instante.

Para que la computadora **recuerde** algo, necesitamos guardarlo en una "caja" y ponerle una etiqueta (nombre).
En programación, a estas cajas con etiqueta las llamamos **VARIABLES**.

Mira este ejemplo:

```python
# 1. Guardamos el número 10 en una caja llamada 'puntaje'
puntaje = 10

# 2. Le decimos a la computadora: "Muestrame qué hay en la caja puntaje"
print(puntaje)
```

**¿Viste?** No imprimió la palabra "puntaje", sino el número **10** que estaba guardado adentro.

Ahora vamos a cambiar lo que hay en la caja. La memoria es variable (por eso se llama Variable).

```python
# Le sumamos 5 puntos extra a lo que ya teniamos
puntaje = puntaje + 5

# Vemos el nuevo resultado
print(puntaje)
```

### 3. Texto vs. Números
Las variables también pueden guardar texto. Pero ojo: para que la computadora sepa que es texto, debemos usar **comillas** `" "`.

* `5` es un número (matemática).
* `"5"` es un dibujo de un cinco (texto).

Vamos a guardar tu nombre en una variable.

```python
# Guardamos texto en la variable 'saludo'
saludo = "Hola, aspirante del ITBA"

print(saludo)
```

### 4. Interactuando
Ahora vamos a combinar todo.
Usaremos una función llamada `input()` que detiene el programa y espera a que escribas algo.
Lo que escribas, se guardará en una variable.

```python
print("El sistema se está iniciando...")

# 1. Creamos la variable 'nombre' y guardamos lo que escribas
nombre = input("Escribe tu nombre aquí: ")

# 2. Usamos esa memoria para saludarte
print("Bienvenido al curso, " + nombre)
```

¡Excelente! Acabas de usar memoria (variables), funciones (print/input) y datos (texto) sin darte cuenta.
